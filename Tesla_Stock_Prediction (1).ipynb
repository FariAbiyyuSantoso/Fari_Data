```python
import yfinance as yf
```

```python
tesla = yf.Ticker("TSLA")
```

```python
tesla = tesla.history(period="max")
```

```python
tesla.plot.line(y="Close", use_index=True)
```

```python
tesla["tomorrow"] = tesla["Close"].shift(-1)
```

```python
tesla
```

```python
tesla["Target"] = (tesla["tomorrow"] > tesla ["Close"]).astype(int)
```

```python
tesla
```

```python
tesla = tesla.loc["2000-01-01":].copy()
```

```python
tesla
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier(n_estimators=100, min_samples_split=100, random_state=1)

train = tesla.iloc[:-100]
test = tesla.iloc[:-100]

predictors = ["Close", "Volume", "Open", "High", "Low"]
model.fit(train[predictors], train["Target"])
```

```python
from sklearn.metrics import precision_score
```

```python
predictions = model.predict(test[predictors])
```

```python
import pandas as pd
predictions = pd.Series(predictions, index=test.index)
```

```python
precision_score(test["Target"], predictions)
```

```python
predictions = pd.Series(predictions, index=test.index, name="Predictions")
combined = pd.concat([test["Target"], predictions], axis=1)
combined.plot()
```

```python
def predict(train, test, predictors, model):
    model.fit(train[predictors], train["Target"])
    predictions = model.predict(test[predictors])
    predictions = pd.Series(predictions, index=test.index, name = "Predictions")
    combined = pd.concat([test["Target"], predictions], axis=1)
    return combined
```

```python
def backtest(data, model, predictors, start=2500, step=250):
    all_predictions = []

    # Iterasi data dengan ukuran step tertentu
    for i in range(start, data.shape[0], step):
        train = data.iloc[0:i].copy()  # Data latih hingga indeks i
        test = data.iloc[i:(i+step)].copy()  # Data uji pada rentang step berikutnya
        
        # Prediksi menggunakan model
        predictions = predict(train, test, predictors, model)
        all_predictions.append(predictions)
    
    # Gabungkan semua prediksi menjadi DataFrame
    return pd.concat(all_predictions)
```

```python
predictions = backtest(tesla, model, predictors)
```

```python
predictions["Predictions"].value_counts()
```

```python
precision_score(predictions["Target"], predictions["Predictions"])
```

```python
predictions["Target"].value_counts() / predictions.shape[0]
```

```python
horizons = [2, 5, 60, 250, 1000]

new_predictors = []

for horizon in horizons:
    rolling_averages = tesla.rolling(horizon).mean()
    
    ratio_column = f"Close Ratio_{horizon}"
    tesla[ratio_column] = tesla["Close"] / rolling_averages["Close"]
    
    trend_column = f"Trend_{horizon}"
    tesla[trend_column] = tesla.shift(1).rolling(horizon).sum()["Target"]
    
    new_predictors += [ratio_column, trend_column]
```

```python
type(tesla)
```

```python
model = RandomForestClassifier(n_estimators=200, min_samples_split=50, random_state=1)
```

```python
def predict(train, test, predictors, model): 
        model.fit(train[predictors], train["Target"])
        predictions = model.predict_proba(test[predictors])[:,1]
        predictions[predictions >= .6] = 1
        predictions[predictions < .6] = 0
        predictions = pd.Series(predictions, index=test.index, name = "Predictions")
        combined = pd.concat([test["Target"], predictions], axis=1)
        return combined
```

```python
predictions = backtest(tesla, model, new_predictors)
```

```python
predictions["Predictions"].value_counts()
```

```python
precision_score(predictions["Target"], predictions["Predictions"])
```
